```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
path = "../dataset/diabetes_binary_5050split_health_indicators_BRFSS2023.csv"

df = pd.read_csv(path)

# Target
y = df.iloc[:, 0].values

# Features (same as dataset[:, 1:-2])
X = df.iloc[:, 1:].values
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
pca = PCA(n_components=0.95)
X_pca = pca.fit_transform(X_scaled)

print("Original dim:", X.shape[1])
print("PCA dim:", X_pca.shape[1])
print("Explained variance:", pca.explained_variance_ratio_.sum())
```

```python
display(pd.DataFrame(X_pca))
```

```python
feature_names = df.columns[1:]  # same slice as X

loadings = pd.DataFrame(
    pca.components_.T,
    index=feature_names,
    columns=[f"PC{i+1}" for i in range(pca.n_components_)]
)

loadings
```

```python
importance = (loadings**2).sum(axis=1)
importance = importance.sort_values(ascending=False)

importance
```

```python
from sklearn.feature_selection import mutual_info_classif

mi = mutual_info_classif(X, y, discrete_features=True)
pd.Series(mi, index=feature_names).sort_values(ascending=False)
```
