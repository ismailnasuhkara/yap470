## Without Dropouts or L1 Regularization

```python
NNBase_array = []
for i in range(0,10):
    NNBase = NeuralNetwork()
    NNBase.build_model(input_dim=22, dropout=False, regularize=False)
    NNBase_array.append(NNBase)
```

```python
for i in range(0,10):
    NNBase_array[i].load_data("../dataset/dataset_preprocessed.csv")
```

```python
for i in range(0,10):
    print(f"----- MODEL {i+1} TRAINING -----")
    NNBase_array[i].train(10)
```

```python
for i in range(0,10):
    print(f"----- MODEL {i+1} EVALUATION -----")
    NNBase_array[i].evaluate()
```
