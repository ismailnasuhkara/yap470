```python
from neural_network import NeuralNetwork
from sklearn.metrics import precision_score, recall_score, f1_score, confusion_matrix
```

```python
NNs = []
for i in range(0,10):
    NNs.append(NeuralNetwork())
```

```python
for i in range(0,10):
    NNs[i].load_data("../dataset/diabetes_binary_5050split_health_indicators_BRFSS2023.csv")
```

```python
for i in range(0,10):
    print(f"----- TRAINING NN {i+1} -----")
    NNs[i].train()
```

```python
results = []
for i in range(0,10):
    o1, o2, o3, o4 = NNs[i].evaluate()
    results_tuple = (o1,o2,o3,o4)
    results.append(results_tuple)
```

```python
for i in range(0,10):
    display(results[i][0][["Threshold","Accuracy","F1-Score","Recall"]].sort_values(by="F1-Score",ascending=False))
```

```python
for i in range(0,10):
    result = results[i]
    NNs[i].plot_threshold_analysis(result[0],result[1],result[2])
```
